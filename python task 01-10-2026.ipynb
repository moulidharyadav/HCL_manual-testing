```python
def count_numbers(transactions, target):
    prefix_sum = 0
    count = 0
    prefix_map = {0: 1}
    for amount in transactions:
        prefix_sum += amount
        if prefix_sum - target in prefix_map:
            count += prefix_map[prefix_sum - target]
        prefix_map[prefix_sum] = prefix_map.get(prefix_sum, 0) + 1
    return count
transactions = [1, 2, 3, -2, 5]
target = 3
print(count_numbers(transactions, target))
```

```python
def has_cycle(n, graph):
    state = [0] * n

    def dfs(node):
        state[node] = 1

        for next_node in graph[node]:

            if state[next_node] == 1:
                return True

            if state[next_node] == 0:
                if dfs(next_node):
                    return True

        state[node] = 2
        return False

    for i in range(n):
        if state[i] == 0:
            if dfs(i):
                return True

    return False


n = 4

graph = [
    [1],
    [2],
    [3],
    [1]
]

print(has_cycle(n, graph))
```
